```python
import sqlite3
import requests

# Danh sách quốc gia
country_list = [
    ("NZ", "New Zealand"),
    ("AU", "Australia"),
    ("US", "United States"),
    ("CA", "Canada"),
    ("GB", "United Kingdom")
]

# Kết nối DB
conn = sqlite3.connect("universities.db")
cur = conn.cursor()
```

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS countries (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    code TEXT UNIQUE NOT NULL,
    name TEXT NOT NULL,
    flag_url TEXT
)
""")

cur.execute("""
CREATE TABLE IF NOT EXISTS universities_basic (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    country_id INTEGER,
    state TEXT,
    domain TEXT,
    website TEXT,
    FOREIGN KEY(country_id) REFERENCES countries(id)
)
""")

conn.commit()

for code, name in country_list:
    cur.execute(
        "INSERT OR IGNORE INTO countries (code, name, flag_url) VALUES (?, ?, ?)",
        (code.lower(), name, f"https://flags.com/{code.lower()}.png")
    )
conn.commit()
```

```python
import json
import time

def safe_get_json(url, retries=3):
    for attempt in range(retries):
        try:
            res = requests.get(url, timeout=10)

            # check status
            if res.status_code != 200:
                print("HTTP error:", res.status_code)
                continue

            # check content-type
            if "application/json" not in res.headers.get("Content-Type", ""):
                print("Not JSON:", res.headers.get("Content-Type"))
                continue

            # Try parse JSON
            return res.json()

        except Exception as e:
            print(f"JSON parse failed (attempt {attempt+1}):", e)
            time.sleep(1)

    return []   # return empty list if fail
```

```python
for code, name in country_list:
    print(f"Fetching: {name} ...")

    url = f"http://universities.hipolabs.com/search?country={name.replace(' ', '+')}"

    data = safe_get_json(url)

    if not data:
        print(f"!! Failed to fetch JSON for {name}, skipping...")
        continue

    # Lấy id quốc gia
    cur.execute("SELECT id FROM countries WHERE code = ?", (code.lower(),))
    country_id = cur.fetchone()[0]

    for uni in data:
        cur.execute("""
            INSERT INTO universities_basic (name, country_id, state, domain, website)
            VALUES (?, ?, ?, ?, ?)
        """, (
            uni.get("name"),
            country_id,
            uni.get("state-province"),
            (uni.get("domains") or [None])[0],
            (uni.get("web_pages") or [None])[0]
        ))

    conn.commit()

print("DONE")
```
